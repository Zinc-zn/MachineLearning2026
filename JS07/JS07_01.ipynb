# JS07 - Klasterisasi: Lab 1 (K-Means pada Dataset Iris)

Praktikum pembuatan model **K-Means** dengan kasus sederhana menggunakan dataset **Iris**. Dataset ini sebetulnya telah memiliki label (dapat digunakan untuk klasifikasi), namun pada praktikum ini kita berasumsi dataset **tidak memiliki label** sehingga relevan untuk proses klasterisasi.


## Langkah 0 - Download Dataset

Dataset `Iris.csv` berada pada folder yang sama dengan notebook ini (diunduh dari halaman modul).

## Langkah 1 - Import Library

```python
# Import required library
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

## Langkah 2 - Load dan Inspeksi Data

```python
df = pd.read_csv('Iris.csv')
df.head()
```

```python
df.info()
```

Terdapat **150 data** dan tidak ada *missing values* pada seluruh variabel.

## Langkah 3 - Seleksi Fitur

Karena hanya membutuhkan fitur untuk proses klasterisasi, pisahkan fitur dan label dengan *slicing*. Slicing `[:, 1:-1]` berarti kolom `Id` dan `Species` (label) tidak digunakan.

```python
# Features Selection
X = df.iloc[:, 1:-1]  # slice the features
y = df.iloc[:, -1]    # this is the labels (we don't need it)

print(X.head())
```

## Langkah 4 - Plotting

Plot sebaran data berdasarkan 2 fitur, yaitu `SepalLengthCm` dan `SepalWidthCm`.

```python
# Make a scatterplot using
# SepalLengthCm and SepalWidthCm
plt.xlabel('SepalLengthCm')
plt.ylabel('SepalWidthCm')
plt.scatter(X.iloc[:, 0], X.iloc[:, 1], s=100)
```

Secara visual, kita mungkin dapat berasumsi bahwa terdapat **dua cluster** pada data.

## Langkah 5 - Membuat Model K-Means

Buat model K-Means dengan asumsi terdapat 2 cluster sesuai amatan visual pada scatterplot.

> **Catatan:** `random_state=42` ditambahkan agar hasil klasterisasi (dan nilai SSE) dapat direproduksi, karena posisi centroid awal ditentukan secara acak.

```python
# Build a K-Means Model
# Based on visual inspection using scatterplot,
# we can assumpt that the data has 2 cluster
# (even though, from the label, there are 3 labels in this data)

from sklearn.cluster import KMeans

# Initiate K-Means object with number of cluster is 2
cl_kmeans = KMeans(n_clusters=2, random_state=42, n_init=10)

# Fit and predict
y_kmeans = cl_kmeans.fit_predict(X)
```

```python
# Plot the clustering result
plt.scatter(X.iloc[:, 0], X.iloc[:, 1], s=100, c=y_kmeans)

# Plot the centroid
centers = cl_kmeans.cluster_centers_
plt.scatter(centers[:, 0], centers[:, 1], c='red', s=200, alpha=0.5)
```

## Langkah 6 - Evaluasi Model

Nilai metrik evaluasi yang digunakan adalah ***sum of square error* (SSE)**. Nilai SSE didapatkan dengan mengakses properti `inertia_` dari objek K-Means yang telah dilatih.

```python
# Print SSE value
print(f'Nilai SSE: {cl_kmeans.inertia_}')
```

**NB:** Nilai SSE dapat berbeda apabila `random_state` berbeda. Ingat, titik centroid awal ditentukan pada lokasi yang acak!

## Langkah 7 - Menentukan k Terbaik dengan Elbow Method

Cara kerja elbow method:

1. Tentukan rentang k yang ingin dievaluasi.
2. Kalkulasi nilai SSE untuk setiap nilai k pada rentang yang telah ditentukan.
3. Buat grafik *lineplot* untuk mendapatkan visualisasi sebagai dasar justifikasi.

```python
# Find the best k
# using Elbow Method

# List SSE values
sse = []

# Build k range from 1 to 9
K = range(1, 10)

# Find the SSE value for each k
for k in K:
    kmeanModel = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeanModel.fit(X)
    sse.append(kmeanModel.inertia_)

# Plotting the distortions
plt.figure(figsize=(8, 4))
plt.plot(K, sse, "bx-")
plt.xlabel("k")
plt.ylabel("SSE")
plt.title("Elbow Method to Find The Best K")
plt.show()
```

**Analisis:** Nilai SSE pasti semakin mengecil seiring bertambahnya jumlah cluster - jumlah cluster terbaik **bukan** karena nilai SSE yang paling rendah. Kita harus menemukan posisi di mana nilai k telah turun dengan tajam diikuti penurunan yang lebih stabil (posisi "elbow"). Pada konteks fitur yang digunakan, **k=3** menjadi nilai yang paling optimal: pada k=2 nilai SSE masih terlalu tinggi dibanding k=4, sedangkan k yang besar (misal k=8) hanya menghasilkan penurunan kecil yang tidak berarti.

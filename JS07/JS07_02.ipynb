# JS07 - Klasterisasi: Lab 2 (Klasterisasi Data Non-Linear dengan DBSCAN)

Pada praktikum sebelumnya, data dapat dipisahkan secara linier. Bagaimana jika data tidak dalam bentuk sebaran yang dapat dipisahkan secara linier? KMeans tidak dapat melakukan tugas ini dengan baik, sehingga digunakan algoritma **DBSCAN** untuk melakukan klasterisasi pada data non-linear.


## Langkah 0 - Import Library

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

## Langkah 1 - Membuat Dataset Sintetis

Data sintetis dibuat berbentuk **spiral (moons)** untuk menyimulasikan kondisi non-linear.

```python
from sklearn.datasets import make_moons

X, y = make_moons(200, noise=.05, random_state=0)
```

```python
plt.scatter(X[:, 0], X[:, 1])
```

Jika diamati secara visual, terdapat dua cluster di dalam dataset. Namun apakah KMeans dapat bekerja dengan baik dalam kasus ini? Mari kita buktikan.

## Langkah 2 - Klasterisasi dengan K-Means

```python
from sklearn.cluster import KMeans

kmeans = KMeans(2, random_state=0, n_init=10)
cluster = kmeans.fit_predict(X)

plt.scatter(X[:, 0], X[:, 1], c=cluster, s=50, cmap='viridis')
```

Terdapat beberapa data yang tidak terklasterisasi secara tepat - mudah dibayangkan garis imajiner linier yang digunakan untuk memisahkan 2 cluster data. Lakukan evaluasi lebih lanjut dengan *silhouette coefficient*.

```python
from sklearn.metrics import silhouette_score

print(silhouette_score(X, cluster))
```

Nilai ini menunjukkan bahwa data **tidak terklasterisasi dengan cukup baik**.

## Langkah 3 - Klasterisasi dengan DBSCAN

***Ingat, DBSCAN sangat sensitif terhadap parameter epsilon dan minimal sampel!***

```python
from sklearn.cluster import DBSCAN

dbscan = DBSCAN(eps=0.3, min_samples=2)
cluster_db = dbscan.fit_predict(X)

plt.scatter(X[:, 0], X[:, 1], c=cluster_db, s=50, cmap='viridis')
```

Berdasarkan scatterplot, data dapat terklasterisasi dengan sangat baik. Untuk memastikannya, lakukan evaluasi kembali dengan *silhouette score*.

```python
from sklearn.metrics import silhouette_score

print(silhouette_score(X, cluster_db))
```

Hasil silhouette score menunjukkan hasil yang tidak optimal, padahal cluster terbentuk dengan baik secara visual. Mengapa demikian? **Silhouette coefficient berasumsi cluster dapat terpisah dengan baik dalam bentuk linier**, sehingga jika bentuk distribusi data acak/non-linier, evaluasi silhouette coefficient tidak dapat bekerja dengan baik.

Evaluasi sederhana yang dapat dilakukan:

1. Prosentase data noise
2. Inspeksi visual

Secara metrik dapat menggunakan **Davies-Bouldin Index (DBI)** - fokus pada separasi indeks, **semakin kecil semakin baik**, rentang 0 hingga tak hingga.

```python
from sklearn.metrics import davies_bouldin_score

# Compute and print Davies-Bouldin Index
db = davies_bouldin_score(X, cluster_db)

print(f"Davies-Bouldin Index: {db}")
```

```python
# Persentase data noise (label -1)
n_noise = list(cluster_db).count(-1)
print(f'Jumlah noise : {n_noise} dari {len(cluster_db)} data ({n_noise/len(cluster_db):.2%})')
print(f'Jumlah cluster (tanpa noise): {len(set(cluster_db)) - (1 if -1 in cluster_db else 0)}')
```

**Kesimpulan Lab 2:** DBSCAN dengan `eps=0.3` dan `min_samples=2` mampu membentuk 2 cluster sesuai bentuk data (visual memisahkan kedua bulan dengan sempurna, tanpa noise). Evaluasi siluet yang rendah tidak mencerminkan hasil klasterisasi yang buruk karena asumsi ke-linieran metrik tersebut tidak terpenuhi pada data non-linier; DBI yang relatif kecil dan persentase noise yang nol menjadi indikator yang lebih tepat untuk kasus ini.

## Langkah 1 - Import Library

```python
# import library
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
import seaborn as sns
```

## Langkah 2 - Membuat Data Dummy

Model yang digunakan adalah generative model sederhana untuk menentukan label pada poin baru. Kasus pada data dummy ini memperlihatkan dua kelas yang telah terpisah.

```python
from sklearn.datasets import make_blobs
X, y = make_blobs(n_samples=50, centers=2,
                  random_state=0, cluster_std=0.60)
plt.scatter(X[:, 0], X[:, 1], c=y, s=50, cmap='autumn')
```

## Langkah 3 - Ilustrasi Garis Pemisah

Untuk membuat model klasifikasi, diperlukan garis pemisah data dari dua himpunan. Namun masih tersisa kasus di mana garis pemisah yang memenuhi syarat lebih dari satu - seperti diilustrasikan berikut.

```python
xfit = np.linspace(-1, 3.5)
plt.scatter(X[:, 0], X[:, 1], c=y, s=50, cmap='autumn')

for m, b in [(1, 0.65), (0.5, 1.6), (-0.2, 2.9)]:
    plt.plot(xfit, m * xfit + b, '-k')

plt.xlim(-1, 3.5)
```

Banyaknya garis pemisah menunjukkan munculnya beberapa kemungkinan suatu data baru masuk ke dalam kategori tertentu. Sebuah data baru $x$ akan dikategorikan berbeda-beda tergantung garis mana yang dipilih.

## Langkah 4 - Ilustrasi Margin

Margin dalam suatu garis dapat digunakan untuk memperjelas posisi sebuah data terhadap label.

```python
xfit = np.linspace(-1, 3.5)
plt.scatter(X[:, 0], X[:, 1], c=y, s=50, cmap='autumn')

for m, b, d in [(1, 0.65, 0.33), (0.5, 1.6, 0.55), (-0.2, 2.9, 0.2)]:
    yfit = m * xfit + b
    plt.plot(xfit, yfit, '-k')
    plt.fill_between(xfit, yfit - d, yfit + d, edgecolor='none',
                     color='#AAAAAA', alpha=0.4)

plt.xlim(-1, 3.5)
```

Margin berfungsi menunjukkan model yang paling optimal. Kriteria optimal diperoleh dari penentuan garis yang mampu **memaksimalkan margin** - inilah inti dari SVM.

## Langkah 5 - Fitting Model

Selanjutnya dilakukan pelatihan model SVM dengan kernel linier.

```python
from sklearn.svm import SVC  # "Support vector classifier"
model = SVC(kernel='linear', C=1E10)
model.fit(X, y)
```

Buat fungsi bantu untuk memvisualisasikan hasil fitting.

```python
# buat sebuah fungsi untuk menampilkan fitting data

def plot_svc_decision_function(model, ax=None, plot_support=True):

    if ax is None:
        ax = plt.gca()
    xlim = ax.get_xlim()
    ylim = ax.get_ylim()

    # buat grid untuk evaluasi model
    x = np.linspace(xlim[0], xlim[1], 30)
    y = np.linspace(ylim[0], ylim[1], 30)
    Y, X = np.meshgrid(y, x)
    xy = np.vstack([X.ravel(), Y.ravel()]).T
    P = model.decision_function(xy).reshape(X.shape)

    # plot batas dan margin
    ax.contour(X, Y, P, colors='k',
               levels=[-1, 0, 1], alpha=0.5,
               linestyles=['--', '-', '--'])

    # plot support vectors
    if plot_support:
        ax.scatter(model.support_vectors_[:, 0],
                   model.support_vectors_[:, 1],
                   s=300, linewidth=1, facecolors='none');
    ax.set_xlim(xlim)
    ax.set_ylim(ylim)
```

Lakukan plotting dengan fungsi yang telah dibuat.

```python
plt.scatter(X[:, 0], X[:, 1], c=y, s=50, cmap='autumn')
plot_svc_decision_function(model)
```

Beberapa titik data yang menyentuh (masuk ke area) garis disebut sebagai **support vector**. Untuk mengetahui titik mana saja yang menjadi support vector, panggil atribut `support_vectors_`.

```python
model.support_vectors_
```

Hasil fitting merupakan kunci keberhasilan suatu classifier (yang hanya dilihat dari posisi support vector). Data lain yang berada jauh dari margin dan tetap di area yang tepat tidak memodifikasi fitting. Untuk membuktikannya, bandingkan hasil fitting saat memakai 60 dan 120 data awal.

```python
def plot_svm(N=10, ax=None):
    X, y = make_blobs(n_samples=200, centers=2,
                      random_state=0, cluster_std=0.60)
    X = X[:N]
    y = y[:N]
    model = SVC(kernel='linear', C=1E10)
    model.fit(X, y)

    ax = ax or plt.gca()
    ax.scatter(X[:, 0], X[:, 1], c=y, s=50, cmap='autumn')
    ax.set_xlim(-1, 4)
    ax.set_ylim(-1, 6)
    plot_svc_decision_function(model, ax)

fig, ax = plt.subplots(1, 2, figsize=(16, 6))
fig.subplots_adjust(left=0.0625, right=0.95, wspace=0.1)
for axi, N in zip(ax, [60, 120]):
    plot_svm(N, axi)
    axi.set_title('N = {0}'.format(N))
```

Berdasarkan hasil di atas, **perubahan jumlah data tidak mempengaruhi model** - keputusan model sepenuhnya ditentukan oleh support vector-nya saja.

Ilustrasi interaktif (jumlah data dapat dipilih antara 10 atau 200):

```python
# jumlah data dapat dipilih di antara 10 atau 200 buah data, terlihat tidak ada perubahan pada model
# (catatan: butuh ipywidgets; jalankan notebook secara interaktif di Jupyter/VS Code untuk slider)
from ipywidgets import interact, fixed
interact(plot_svm, N=[10, 200], ax=fixed(None))
```

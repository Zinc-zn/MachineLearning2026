```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler

# Load dataset
df_wbc = pd.read_csv('wbc.csv')

# 1. Pisahkan variabel yang dapat digunakan dan tidak
df_wbc = df_wbc.drop(columns=['id', 'Unnamed: 32'])

# 2. Proses encoding pada kolom 'diagnosis' (M=1, B=0)
le = LabelEncoder()
df_wbc['diagnosis'] = le.fit_transform(df_wbc['diagnosis'])

# 3. Standardisasi kolom numerik (selain diagnosis)
std = StandardScaler()
numerical_cols = df_wbc.columns.drop('diagnosis')
df_wbc[numerical_cols] = std.fit_transform(df_wbc[numerical_cols])

# Tampilkan hasil
df_wbc.head()
```

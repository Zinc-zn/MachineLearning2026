# 0. Set up data

```python
import pandas as pd
from sklearn.model_selection import train_test_split

file_path = '/content/wbc.csv'
df = pd.read_csv(file_path)
```

# 1. Pisahkan antara variabel yang dapat digunakan dan variabel yang tidak dapat digunakan.

```python
X = df.drop(columns=['id', 'diagnosis'])
if 'Unnamed: 32' in X.columns:
    X = X.drop(columns=['Unnamed: 32'])

y = df['diagnosis']

display(X.head())
print(f'Shape of X: {X.shape}')
```

# 2. Lakukan proses encoding pada kolom "diagnosis".

```python
from sklearn.preprocessing import LabelEncoder

# 2. Lakukan proses encoding pada kolom "diagnosis"
le = LabelEncoder()
y_encoded = le.fit_transform(y)

print(f'Original classes: {le.classes_}')
print(f'Encoded sample: {y_encoded[:5]}')
```

# 3. Lakukan proses standardisasi pada semua kolom yang memiliki nilai numerik.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_std = scaler.fit_transform(X)

# Menampilkan hasil standardisasi (5 baris pertama)
print("Hasil Standardisasi (5 baris pertama):")
print(X_std[:5])
```

# 4. Lakukan proses seleksi fitur. Anda dapat menggunakan SelectKBest.

```python
from sklearn.feature_selection import SelectKBest, f_classif

# Kita akan memilih 10 fitur terbaik sebagai contoh
selector = SelectKBest(score_func=f_classif, k=10)
X_selected = selector.fit_transform(X_std, y_encoded)

print(f"Ukuran data setelah seleksi fitur: {X_selected.shape}")
```

# 5. Lakukan proses pengujian dengan model Logistic Regression seperti pada praktikum 1.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

X_train, X_test, y_train, y_test = train_test_split(X_selected, y_encoded, test_size=0.2, random_state=42)

model = LogisticRegression()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)

print(f"Akurasi Model Logistic Regression: {accuracy:.4f}")
```

# 6. Anda dapat menggunakan model pipeline untuk mempermudah perkejaan Anda.

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.linear_model import LogisticRegression

# Membagi data menjadi train dan test set
X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, test_size=0.2, random_state=42)

pipeline = Pipeline([
    ('scaler', StandardScaler()), # 3. Standardisasi
    ('feature_selection', SelectKBest(score_func=f_classif, k=10)), # 4. Seleksi Fitur (misal k=10)
    ('classifier', LogisticRegression()) # 5. Logistic Regression
])

pipeline.fit(X_train, y_train)

# Evaluasi
accuracy = pipeline.score(X_test, y_test)
print(f'Accuracy with k=10: {accuracy:.4f}')
```

# 7. Berdasarkan hasil analisa Anda, berapa jumlah fitur terbaik yang dapat digunakan? Apa saja fitur tersebut?

```python
import numpy as np

selector = pipeline.named_steps['feature_selection']
mask = selector.get_support()
features = X.columns[mask]

print(f'Jumlah fitur terbaik yang digunakan: {len(features)}')
print('Fitur-fitur tersebut adalah:')
for f in features:
    print(f'- {f}')
```
